```python
import os
from groq import Groq


API_KEY = os.getenv("GROQAI_API_KEY")

# Initialize Groq Client
client = Groq(api_key=API_KEY)

def generate_groq_response(prompt, model="mixtral-8x7b-32768"):
    """
    Sends a prompt to Groq AI and returns the generated response.
    """
    try:
        chat_completion = client.chat.completions.create(
            messages=[
                {"role": "system", "content": "You are a helpful assistant."},
                {"role": "user", "content": prompt}
            ],
            model=model  # Available models: "llama-3.3-70b-versatile", "mixtral-8x7b-32768"
        )

        return chat_completion.choices[0].message.content

    except Exception as e:
        print(f"Error: {e}")
        return None

prompt = "Explain quantum computing in simple terms."
response = generate_groq_response(prompt)
print("Groq AI Response:", response)
```

```python
API_KEY = ""
client = Groq(api_key=API_KEY)

def get_answer(messages):
    """
    Uses Groq AI to generate a response based on user messages.
    """
    system_message = [{"role": "system", "content": "You are a helpful AI chatbot that answers user questions."}]

    if isinstance(messages, str):
        messages = [{"role": "user", "content": messages}]

    messages = system_message + messages  # Combine system and user messages

    try:
        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",  # Other option: "mixtral-8x7b"
            messages=messages
        )

        return response.choices[0].message.content

    except Exception as e:
        print(f"Error calling Groq AI: {e}")
        return "I'm sorry, but I couldn't process your request."

messages = "What is Data wrangling.Write in 3 sentence"
get_answer(messages)
```
